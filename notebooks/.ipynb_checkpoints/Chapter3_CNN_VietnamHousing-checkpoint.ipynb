## 1. Load and preprocess the data

We drop the free-text `Address` column and encode the categorical columns (`House direction`, `Balcony direction`, `Legal status`, `Furniture state`) as integer codes, treating the remaining columns as a short 1D feature sequence, exactly as the diabetes notebook does for its eight features.

```python
import numpy as np
import pandas as pd

np.random.seed(42)

DATA_PATH = "../data/vietnam_housing_dataset.csv"
df = pd.read_csv(DATA_PATH)
print(df.shape)
df.head()
```

```python
NUM_COLS = ['Area', 'Frontage', 'Access Road', 'Floors', 'Bedrooms', 'Bathrooms']
CAT_COLS = ['House direction', 'Balcony direction', 'Legal status', 'Furniture state']
TARGET_COL = 'Price'

# This dataset has a lot of missing values (see slide's tabular-data warning --
# real-world tabular data is messy). We impute numeric columns with the median
# and treat missing categories as their own "Unknown" category.
for c in NUM_COLS:
    df[c] = df[c].fillna(df[c].median())
for c in CAT_COLS:
    df[c] = df[c].fillna('Unknown').astype('category').cat.codes

# Drop a few extreme price outliers (top 1%) so the loss isn't dominated by them.
df = df[(df[TARGET_COL] > 0) & (df[TARGET_COL] < df[TARGET_COL].quantile(0.99))].reset_index(drop=True)

FEATURE_COLS = NUM_COLS + CAT_COLS
print("Features:", FEATURE_COLS, " (", len(FEATURE_COLS), "columns )")

X = df[FEATURE_COLS].values.astype(np.float64)
y = df[TARGET_COL].values.astype(np.float64)

# Standardize inputs; also standardize the target for stable training,
# and remember mu/sigma to convert predictions back to the price scale.
X_mu, X_sigma = X.mean(axis=0), X.std(axis=0) + 1e-8
X = (X - X_mu) / X_sigma

y_mu, y_sigma = y.mean(), y.std()
y_scaled = (y - y_mu) / y_sigma

n_train = int(0.8 * len(X))
idx = np.random.permutation(len(X))
X, y, y_scaled = X[idx], y[idx], y_scaled[idx]
X_train, X_test = X[:n_train], X[n_train:]
y_train, y_test = y[:n_train], y[n_train:]                    # original price scale (for reporting)
y_train_s, y_test_s = y_scaled[:n_train], y_scaled[n_train:]   # standardized (for training)

print("X_train:", X_train.shape, " X_test:", X_test.shape)
```

```python
X_train_seq = X_train[:, None, :]   # (N, 1, num_features)
X_test_seq = X_test[:, None, :]
X_train_seq.shape
```

## 2. NumPy from scratch

Identical layer implementations to the diabetes notebook (`Conv1D`, `ReLU`, `MaxPool1D`, `Flatten`, `Dense`), but the output layer is **linear** (no sigmoid) and the loss is **MSE**, per the regression guidance in the slides.


```python
class Conv1D:
    def __init__(self, in_channels, out_channels, kernel_size, seed=0):
        rng = np.random.default_rng(seed)
        scale = np.sqrt(2.0 / (in_channels * kernel_size))
        self.W = rng.normal(0, scale, size=(out_channels, in_channels, kernel_size))
        self.b = np.zeros(out_channels)

    def forward(self, x):
        batch, in_c, length = x.shape
        out_c, _, k = self.W.shape
        out_len = length - k + 1
        out = np.zeros((batch, out_c, out_len))
        for i in range(out_len):
            window = x[:, :, i:i + k]
            out[:, :, i] = np.tensordot(window, self.W, axes=([1, 2], [1, 2])) + self.b
        self.cache = (x, out_len)
        return out

    def backward(self, dout, lr):
        x, out_len = self.cache
        batch = x.shape[0]
        out_c, in_c, k = self.W.shape
        dW = np.zeros_like(self.W)
        db = np.zeros_like(self.b)
        dx = np.zeros_like(x)
        for i in range(out_len):
            window = x[:, :, i:i + k]
            dW += np.einsum('bo,bik->oik', dout[:, :, i], window)
            db += dout[:, :, i].sum(axis=0)
            dx[:, :, i:i + k] += np.einsum('bo,oik->bik', dout[:, :, i], self.W)
        self.W -= lr * dW / batch
        self.b -= lr * db / batch
        return dx


class ReLU:
    def forward(self, x):
        self.mask = x > 0
        return x * self.mask

    def backward(self, dout, lr=None):
        return dout * self.mask


class MaxPool1D:
    def __init__(self, size=2):
        self.size = size

    def forward(self, x):
        batch, c, length = x.shape
        out_len = length // self.size
        out = np.zeros((batch, c, out_len))
        self.mask = np.zeros_like(x)
        for i in range(out_len):
            seg = x[:, :, i * self.size:(i + 1) * self.size]
            idx = np.argmax(seg, axis=2)
            out[:, :, i] = np.max(seg, axis=2)
            b_idx, c_idx = np.indices((batch, c))
            self.mask[b_idx, c_idx, i * self.size + idx] = 1
        return out

    def backward(self, dout, lr=None):
        batch, c, out_len = dout.shape
        dx = np.zeros_like(self.mask)
        for i in range(out_len):
            seg_mask = self.mask[:, :, i * self.size:(i + 1) * self.size]
            dx[:, :, i * self.size:(i + 1) * self.size] = seg_mask * dout[:, :, i:i + 1]
        return dx


class Flatten:
    def forward(self, x):
        self.shape = x.shape
        return x.reshape(x.shape[0], -1)

    def backward(self, dout, lr=None):
        return dout.reshape(self.shape)


class Dense:
    def __init__(self, in_dim, out_dim, seed=1):
        rng = np.random.default_rng(seed)
        self.W = rng.normal(0, np.sqrt(2.0 / in_dim), size=(in_dim, out_dim))
        self.b = np.zeros(out_dim)

    def forward(self, x):
        self.x = x
        return x @ self.W + self.b

    def backward(self, dout, lr):
        dW = self.x.T @ dout / self.x.shape[0]
        db = dout.mean(axis=0)
        dx = dout @ self.W.T
        self.W -= lr * dW
        self.b -= lr * db
        return dx
```

```python
n_features = X_train_seq.shape[2]
kernel_size = 3 if n_features >= 5 else 2   # keep the kernel valid for the sequence length

conv1_np = Conv1D(in_channels=1, out_channels=16, kernel_size=kernel_size, seed=0)
relu1_np = ReLU()
conv2_np = Conv1D(in_channels=16, out_channels=8, kernel_size=kernel_size, seed=1)
relu2_np = ReLU()
pool_np = MaxPool1D(size=2)
flat_np = Flatten()

_out = conv1_np.forward(X_train_seq[:2])
_out = relu1_np.forward(_out)
_out = conv2_np.forward(_out)
_out = relu2_np.forward(_out)
_out = pool_np.forward(_out)
_out = flat_np.forward(_out)
flat_dim = _out.shape[1]
print("Flattened feature size:", flat_dim)

dense_np = Dense(flat_dim, 1, seed=2)   # linear output -- no sigmoid, this is regression
```

```python
def forward_pass(x):
    z1 = conv1_np.forward(x)
    a1 = relu1_np.forward(z1)
    z2 = conv2_np.forward(a1)
    a2 = relu2_np.forward(z2)
    p = pool_np.forward(a2)
    f = flat_np.forward(p)
    return dense_np.forward(f)   # linear output


def backward_pass(preds, yb, lr):
    dlogits = 2 * (preds - yb) / yb.shape[0]   # d(MSE)/d(preds)
    df_ = dense_np.backward(dlogits, lr)
    dp = flat_np.backward(df_, lr)
    da2 = pool_np.backward(dp, lr)
    dz2 = relu2_np.backward(da2, lr)
    da1 = conv2_np.backward(dz2, lr)
    dz1 = relu1_np.backward(da1, lr)
    conv1_np.backward(dz1, lr)


LR = 0.01
EPOCHS = 60
BATCH_SIZE = 64
n = X_train_seq.shape[0]

history_np = []
for epoch in range(EPOCHS):
    perm = np.random.permutation(n)
    Xs, ys = X_train_seq[perm], y_train_s[perm]
    epoch_losses = []
    for i in range(0, n, BATCH_SIZE):
        xb = Xs[i:i + BATCH_SIZE]
        yb = ys[i:i + BATCH_SIZE].reshape(-1, 1)

        preds = forward_pass(xb)
        loss = np.mean((preds - yb) ** 2)
        epoch_losses.append(loss)

        backward_pass(preds, yb, LR)

    history_np.append(np.mean(epoch_losses))
    if epoch % 10 == 0 or epoch == EPOCHS - 1:
        print(f"epoch {epoch:2d}  MSE(scaled) {history_np[-1]:.4f}")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

preds_test_np_scaled = forward_pass(X_test_seq).flatten()
preds_test_np = preds_test_np_scaled * y_sigma + y_mu   # back to original price scale

results_np = {
    "MAE": mean_absolute_error(y_test, preds_test_np),
    "RMSE": mean_squared_error(y_test, preds_test_np) ** 0.5,
    "R2": r2_score(y_test, preds_test_np),
}
print("NumPy-from-scratch test metrics:", {k: round(v, 4) for k, v in results_np.items()})
```

## 3. PyTorch

Same architecture, linear output, `nn.MSELoss`, trained with autograd.


```python
import torch
import torch.nn as nn
import torch.optim as optim

torch.manual_seed(42)

class HousingCNN(nn.Module):
    def __init__(self, kernel_size):
        super().__init__()
        self.conv1 = nn.Conv1d(1, 16, kernel_size=kernel_size)
        self.relu1 = nn.ReLU()
        self.conv2 = nn.Conv1d(16, 8, kernel_size=kernel_size)
        self.relu2 = nn.ReLU()
        self.pool = nn.MaxPool1d(kernel_size=2)
        self.flatten = nn.Flatten()
        self.fc = nn.LazyLinear(1)   # linear output -- regression

    def forward(self, x):
        x = self.relu1(self.conv1(x))
        x = self.relu2(self.conv2(x))
        x = self.pool(x)
        x = self.flatten(x)
        return self.fc(x)


model_pt = HousingCNN(kernel_size)
criterion = nn.MSELoss()
optimizer = optim.Adam(model_pt.parameters(), lr=0.01)

X_train_t = torch.tensor(X_train_seq, dtype=torch.float32)
y_train_t = torch.tensor(y_train_s, dtype=torch.float32).view(-1, 1)
X_test_t = torch.tensor(X_test_seq, dtype=torch.float32)

EPOCHS = 60
BATCH_SIZE = 64
n = X_train_t.shape[0]
history_pt = []

for epoch in range(EPOCHS):
    perm = torch.randperm(n)
    epoch_losses = []
    for i in range(0, n, BATCH_SIZE):
        batch_idx = perm[i:i + BATCH_SIZE]
        xb, yb = X_train_t[batch_idx], y_train_t[batch_idx]

        optimizer.zero_grad()
        preds = model_pt(xb)
        loss = criterion(preds, yb)
        loss.backward()
        optimizer.step()
        epoch_losses.append(loss.item())

    history_pt.append(sum(epoch_losses) / len(epoch_losses))
    if epoch % 10 == 0 or epoch == EPOCHS - 1:
        print(f"epoch {epoch:2d}  MSE(scaled) {history_pt[-1]:.4f}")
```

```python
model_pt.eval()
with torch.no_grad():
    preds_test_pt = model_pt(X_test_t).numpy().flatten() * y_sigma + y_mu

results_pt = {
    "MAE": mean_absolute_error(y_test, preds_test_pt),
    "RMSE": mean_squared_error(y_test, preds_test_pt) ** 0.5,
    "R2": r2_score(y_test, preds_test_pt),
}
print("PyTorch test metrics:", {k: round(v, 4) for k, v in results_pt.items()})
```

## 4. TensorFlow / Keras

Same architecture, `Dense(1)` with no activation (linear) for the output, `loss="mse"`.


```python
import tensorflow as tf

tf.random.set_seed(42)

model_tf = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(n_features, 1)),
    tf.keras.layers.Conv1D(filters=16, kernel_size=kernel_size, activation="relu"),
    tf.keras.layers.Conv1D(filters=8, kernel_size=kernel_size, activation="relu"),
    tf.keras.layers.MaxPooling1D(pool_size=2),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(1),   # linear output -- regression
])

model_tf.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.01), loss="mse")
model_tf.summary()
```

```python
X_train_tf = np.transpose(X_train_seq, (0, 2, 1))   # channels-last
X_test_tf = np.transpose(X_test_seq, (0, 2, 1))

history_tf = model_tf.fit(
    X_train_tf, y_train_s,
    epochs=60, batch_size=64, verbose=0,
)
print("final training loss (scaled MSE):", history_tf.history["loss"][-1])
```

```python
preds_test_tf = model_tf.predict(X_test_tf, verbose=0).flatten() * y_sigma + y_mu

results_tf = {
    "MAE": mean_absolute_error(y_test, preds_test_tf),
    "RMSE": mean_squared_error(y_test, preds_test_tf) ** 0.5,
    "R2": r2_score(y_test, preds_test_tf),
}
print("TensorFlow test metrics:", {k: round(v, 4) for k, v in results_tf.items()})
```

## 5. Compare the three implementations

```python
comparison = pd.DataFrame(
    [results_np, results_pt, results_tf],
    index=["NumPy (from scratch)", "PyTorch", "TensorFlow/Keras"],
)
comparison.round(4)
```

## 6. Discussion

- **Task-appropriate output.** Per the slides' "A Very Important Lesson", the
  output layer here is **linear**, not sigmoid — regression needs an
  unbounded output, unlike the diabetes notebook's probability output.
- **Same caveat as before**: house features are tabular, not spatial, so the
  CNN's "local connectivity" and "weight sharing" don't have an obvious
  physical meaning here (e.g. `Area` next to `Frontage` in the array is an
  arbitrary column order, not a spatial relationship). Treat this as an
  exercise in composing `Conv → ReLU → Pool → Dense` for a regression task,
  not a benchmark that a 1D CNN beats simpler regressors on this data.
- Compare MAE / RMSE / R² across the three implementations above and note any
  differences in training stability or final error.
